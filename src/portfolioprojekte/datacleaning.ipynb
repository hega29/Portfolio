# numerische spalten berechnen: Wir haben 2 numerische Spalten. 
# Diese sind: 'plz' und 'median_betrab_eur'.
# in Deustchland gibt es 5 Ziffern in der Postleitzahl.
# Zur Berinigung der Daten können wir die Postleitzahlen auf 5 Ziffern überprüfen 
# Dafür wird eine Funktion erstellt, die überprüft, ob die Postleitzahl 5 Ziffern hat.

```python

def check_plz(plz):
    if pd.isna(plz):
        return np.nan
    # in string umwandeln und eventuell nachkommastellen entfernen
    s = str(plz).strip()
    if '.' in s:
        s = s.split('.')[0]
    # nur Ziffern behalten falls es Buchstaben gibt
    s = ''.join(filter(str.isdigit, s))
    # Auf länge prüfen und korrigieren
    if len(s) == 5:
        return str(s)
    elif len(s) == 4:
        return '0' + str(s)  # führende Nullen hinzufügen
    else:
        return np.nan  # ungültige Postleitzahl

```

# kontrolliere welche Postleitzahlen ungültig verwofen wurden
Spalte plz als category umwandeln

```python
df['plz'] = df['plz'].apply(check_plz)

ungueltige_plz = df[df['plz'].isna() & df['plz'].notna()] 

print("Ungueltige Postleitzahlen:" , len(ungueltige_plz))
print(ungueltige_plz[['lieferanten_id', 'lieferanten_name', 'plz']])

```

# Spalte 'median_betrag_eur' hat 2 fehlende Werte. 
# Wir können diese Werte durch den Median der Spalte ersetzen, um die Daten zu bereinigen.
# spalte stadt und plz behandeln: Wir können die fehlenden Werte in der Spalte 'stadt' 
# und 'plz' durch den Wert 'unbekannt' ersetzen, um die Daten zu bereinigen.
# Die Spalte offizielle_iban hat 1 fehlenden Wert. 
# Wir können diesen Wert durch den Wert 'Fehlt' ersetzen,
# um die Daten zu bereinigen.

```python

gesamt_median = df['median_betrag_eur'].median()
#df['median_betrag_eur'] = df['median_betrag_eur'].astype(float).float.strip()
df['median_betrag_eur'] = df['median_betrag_eur'].fillna(gesamt_median)

df['stadt'] = df['stadt'].astype(str).str.strip()
df['stadt'] = df['stadt'].str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Unbekannt')
df['plz'] = df['plz'].str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('0')

df['offizielle_iban'] = df['offizielle_iban'].str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Fehlt')
```

```python
# alle Strin Datentyp als category umwandeln
for col in df.columns:
    if df[col].dtype == 'str':
        df[col] = df[col].astype('category')
```
