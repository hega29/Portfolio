# Spalte hat_anomalie bereinigen

```python
# prüfung ob es negative Werte gibt
df[df['hat_anomalie'] < 0]
```

# negative Werte in der Spalte 'hat_anomalie' werden durch nan ersetzt
# und die Spalte als numerisch deklariert

```python

df.loc[df['hat_anomalie'] < 0, 'hat_anomalie'] = np.nan
df['hat_anomalie'] = pd.to_numeric(df['hat_anomalie'], errors='coerce')
```

# Spalte hat_annomalie als integer casten

```python
df['hat_anomalie'] = df['hat_anomalie'].astype('Int64')
```
