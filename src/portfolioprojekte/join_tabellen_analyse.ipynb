# Bibliotheken importieren

```python
import pandas as pd
import numpy as np
```

# Die 3 Dateien lesen

```python
df_lieferanten = pd.read_csv('lieferanten_bereinigt.csv')
df_rechnungen = pd.read_csv('rechnungen_bereinigt.csv')
df_anomalien = pd.read_csv('labels_anomalien_bereinigt.csv')
```

# erste zwei DataFrame Tabellen über die Spalte lieferanten_id joinen

```python

df_joined = pd.merge(df_lieferanten, 
                     df_rechnungen, 
                     left_on='lieferanten_id', 
                     right_on='lieferanten_id',
                     how='inner')
```

# prüfen ob die join funktioniert hat

```python
df_joined.head() 
```

# Das DF df_anomalien über transaktions_id joinen

```python

df_joined = pd.merge(df_joined, df_anomalien, 
                     left_on='transaktions_id', 
                     right_on='transaktions_id', 
                     how='inner')
```

# Die drei df prüfen 

```python
df_joined.head()
```

# Datentyp checken

```python
df_joined.dtypes
```

# die Spalte plz als string umwandeln, da es sich um eine Postleitzahl handelt

```python

df_joined['plz'] = df_joined['plz'].astype(str)
```

# die spalte hat_anomalie als integer umwandeln, da es sich um eine binäre Variable handelt

# Datentyp des Join checken

```python

df_joined['hat_anomalie'] = df_joined['hat_anomalie'].astype('Int64')
```

```python
df_joined.dtypes
```

# prüfe ob es negative werte bei der spalte hat_anomalie gibt

```python

negative_values = df_joined[df_joined['hat_anomalie'] < 0]
print(negative_values.shape[0])
```

# prüfeob es noch fehlende Werte gibt bei der spalte hat_anomalie

```python

missing_values = df_joined[df_joined['hat_anomalie'].isnull()]
print(missing_values.shape[0])
```

# fehlende Werte in der Spalte hat_anomalie mit nan auffüllen
# die spalte als integrer umwandeln
# Fehlende werte in der Spalte hat_anomalie löschen 

```python

df_joined['hat_anomalie'] = df_joined['hat_anomalie'].replace(['', 'nan', 'NaN', 'None', -1, '-1'], np.nan).astype('Int64')

df_joined['hat_anomalie'] = df_joined['hat_anomalie'].astype('Int64')

df_joined = df_joined.dropna(subset=['hat_anomalie'])
```

# prüfen, ob es noch fehlende Werte gibt

```python

df_joined.isna().sum()
```

# alle string spalten as kategorische Variablen umwandeln

```python

for col in df_joined.columns:
    if df_joined[col].dtype == 'str':
        df_joined[col] = df_joined[col].astype('category')
```

# Datentyp prüfen

```python
df_joined.dtypes
```

# lösche eine der spalten mit gleichen Inhalten

```python

column_to_loeschen = ['ist_duplikat_kopie_x', 'ist_iban_abweichung_x', 'ist_betrags_ausreisser_x']
df_joined = df_joined.drop(columns=column_to_loeschen, errors='ignore')
    


```
