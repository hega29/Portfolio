# Datumswerte bereinigen: Wir können die Spalte 'rechnungsdatum' in ein Datumsformat umwandeln, um die Daten für die Analyse der Buchungen nach Datum vorzubereiten.

```python

df['rechnungsdatum'] = df['rechnungsdatum'].astype(str).str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Unbekannt')
```

# faelligskeitsdatum bereinigen: Wir können die Spalte 'faelligkeitsdatum' in ein Datumsformat umwandeln, um die Daten für die Analyse der Buchungen nach Fälligkeitsdatum vorzubereiten.


```python
df['faelligkeitsdatum'] = df['faelligkeitsdatum'].astype(str).str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Unbekannt')
# df['faelligkeitsdatum'] = pd.to_datetime(df['faelligkeitsdatum'],
```

# prüft, ob es fehlende Werte gibt und summiere die

```python
df.isna().sum()
```

# Beträge saubern und in Zahlen umwandlen

```python

df['betrag_eur'] = (
    df['betrag_eur']
    .astype(str)
    .str.replace(' €', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
df['betrag_eur'] = pd.to_numeric(df['betrag_eur'], errors='coerce')

```

# ungueltige Werte in der spalte 'betrag_eur' behandeln: 
# Wir können die ungueltigen Werte in der Spalte 'betrag_eur' durch den Median der Spalte ersetzen, um die Daten zu bereinigen.

```python

df['betrag_eur'] = df['betrag_eur'].fillna(df['betrag_eur'].median())
```

```python
df.isna().sum()
```

# Daten für die Kostenstellenanalyse vorbereiten: Wir können die Spalte 'kostenstelle' in eine Kategorie umwandeln, 
# um die Daten für die Kostenstellenanalyse vorzubereiten. und leerzeichen entfernen

```python

df['kostenstelle'] = df['kostenstelle'].astype(str).str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Unbekannt').astype('category')
```

# Analyse der spalte buchender_mitarbeiter: 
# Wir können die Spalte 'buchender_mitarbeiter' in eine Kategorie umwandeln,
# um die Daten für die Analyse der Buchungen pro Mitarbeiter vorzubereiten. und leerzeichen entfernen

```python

df['buchender_mitarbeiter'] = df['buchender_mitarbeiter'].astype(str).str.strip().replace(['', 'nan', 'NaN', 'None'], np.nan).fillna('Unbekannt').astype('category')
```

```python
df.isna().sum()
```

# string Datentyp als Kategorie umwandeln

```python

for col in df.columns:
    if df[col].dtype == 'str':
        df[col] = df[col].astype('category')
   
```
